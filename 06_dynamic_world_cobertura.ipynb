# 06 · Cobertura terrestre con Dynamic World

**Objetivo:** Crear un mapa de clases de cobertura y probabilidades.

**Datos:** GOOGLE/DYNAMICWORLD/V1.

**Relevancia para política ambiental y social:** Facilita diagnósticos rápidos de uso del suelo y planificación.

**Limitaciones:** Es una clasificación probabilística; debe evaluarse localmente.


```python
# Instalar dependencias en Google Colab
!pip -q install earthengine-api geemap

import ee
import geemap
import datetime

ee.Authenticate()
ee.Initialize(project="TU_PROYECTO_GEE")

# Área de estudio de ejemplo: entorno de Chachapoyas, Amazonas, Perú
# Reemplázala por un polígono, activo de Earth Engine o coordenadas propias.
aoi = ee.Geometry.Point([-77.87, -6.23]).buffer(30000)

Map = geemap.Map()
Map.centerObject(aoi, 9)
```

```python
dw = (
    ee.ImageCollection("GOOGLE/DYNAMICWORLD/V1")
    .filterBounds(aoi)
    .filterDate("2025-01-01","2025-12-31")
)
label = dw.select("label").mode().clip(aoi)
palette = ["419BDF","397D49","88B053","7A87C6","E49635","DFC35A","C4281B","A59B8F","B39FE1"]
Map.addLayer(label, {"min":0,"max":8,"palette":palette}, "Dynamic World")
Map
```
