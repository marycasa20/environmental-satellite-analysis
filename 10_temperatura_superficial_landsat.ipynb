# 10 · Temperatura superficial con Landsat 8/9

**Objetivo:** Estimar temperatura terrestre y localizar puntos calientes.

**Datos:** LANDSAT/LC08/C02/T1_L2 y banda ST_B10.

**Relevancia para política ambiental y social:** Útil para islas de calor, salud urbana y adaptación climática.

**Limitaciones:** La temperatura superficial no equivale a temperatura del aire.


```python
# Instalar dependencias en Google Colab
!pip -q install earthengine-api geemap

import ee
import geemap
import datetime

ee.Authenticate()
ee.Initialize(project="TU_PROYECTO_GEE")

# Área de estudio de ejemplo: entorno de Chachapoyas, Amazonas, Perú
# Reemplázala por un polígono, activo de Earth Engine o coordenadas propias.
aoi = ee.Geometry.Point([-77.87, -6.23]).buffer(30000)

Map = geemap.Map()
Map.centerObject(aoi, 9)
```

```python
landsat = (
    ee.ImageCollection("LANDSAT/LC08/C02/T1_L2")
    .filterBounds(aoi)
    .filterDate("2025-01-01","2025-12-31")
    .filter(ee.Filter.lt("CLOUD_COVER",30))
)
lst = landsat.select("ST_B10").median().multiply(0.00341802).add(149.0).subtract(273.15).clip(aoi)
Map.addLayer(lst, {"min":5,"max":35,"palette":["blue","cyan","yellow","red"]}, "LST °C")
Map
```
