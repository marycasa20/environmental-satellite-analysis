# 14 · Accesibilidad territorial a servicios

**Objetivo:** Calcular distancia euclidiana desde comunidades hacia servicios.

**Datos:** Puntos de ejemplo; reemplazar por escuelas, centros de salud o mercados.

**Relevancia para política ambiental y social:** Visibiliza desigualdades espaciales y ayuda a priorizar inversión social.

**Limitaciones:** La distancia en línea recta no equivale a tiempo de viaje real.


```python
# Instalar dependencias en Google Colab
!pip -q install earthengine-api geemap

import ee
import geemap
import datetime

ee.Authenticate()
ee.Initialize(project="TU_PROYECTO_GEE")

# Área de estudio de ejemplo: entorno de Chachapoyas, Amazonas, Perú
# Reemplázala por un polígono, activo de Earth Engine o coordenadas propias.
aoi = ee.Geometry.Point([-77.87, -6.23]).buffer(30000)

Map = geemap.Map()
Map.centerObject(aoi, 9)
```

```python
# Puntos ilustrativos: reemplazar por tus datos.
services = ee.FeatureCollection([
    ee.Feature(ee.Geometry.Point([-77.87,-6.23]), {"type":"health"}),
    ee.Feature(ee.Geometry.Point([-77.80,-6.20]), {"type":"school"})
])
service_raster = ee.Image().byte().paint(services, 1)
distance = service_raster.fastDistanceTransform(1024).sqrt().multiply(10).clip(aoi)
Map.addLayer(distance, {"min":0,"max":20000,"palette":["green","yellow","red"]}, "Distancia aproximada (m)")
Map.addLayer(services, {"color":"blue"}, "Servicios")
Map
```
