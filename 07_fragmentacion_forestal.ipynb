# 07 · Fragmentación y borde forestal

**Objetivo:** Distinguir núcleo forestal, borde y zonas no forestales.

**Datos:** Dynamic World y operaciones focales.

**Relevancia para política ambiental y social:** Aporta evidencia sobre conectividad y presión sobre hábitats.

**Limitaciones:** La escala y el umbral determinan fuertemente el resultado.


```python
# Instalar dependencias en Google Colab
!pip -q install earthengine-api geemap

import ee
import geemap
import datetime

ee.Authenticate()
ee.Initialize(project="TU_PROYECTO_GEE")

# Área de estudio de ejemplo: entorno de Chachapoyas, Amazonas, Perú
# Reemplázala por un polígono, activo de Earth Engine o coordenadas propias.
aoi = ee.Geometry.Point([-77.87, -6.23]).buffer(30000)

Map = geemap.Map()
Map.centerObject(aoi, 9)
```

```python
dw = ee.ImageCollection("GOOGLE/DYNAMICWORLD/V1").filterBounds(aoi).filterDate("2025-01-01","2025-12-31")
forest = dw.select("label").mode().eq(1).clip(aoi)
core = forest.focalMin(radius=100, units="meters")
edge = forest.And(core.Not()).selfMask()

Map.addLayer(forest.selfMask(), {"palette":["228B22"]}, "Bosque")
Map.addLayer(core.selfMask(), {"palette":["006400"]}, "Núcleo forestal")
Map.addLayer(edge, {"palette":["orange"]}, "Borde")
Map
```
