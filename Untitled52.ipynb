```python
# ============================================================
# IMAGEN REAL SENTINEL-2 RGB
# CHACHAPOYAS, AMAZONAS, PERÚ
# Fuente: EOX Sentinel-2 Cloudless 2025
# ============================================================

!pip -q install requests pillow matplotlib

import requests
from PIL import Image, ImageDraw, ImageFont
from io import BytesIO
from pathlib import Path
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# 1. CONFIGURACIÓN DEL ÁREA
# ------------------------------------------------------------

# Área aproximada alrededor de Chachapoyas, Amazonas
# Orden WMS: oeste, sur, este, norte
BBOX = "-78.20,-6.60,-77.50,-5.90"

WIDTH = 1600
HEIGHT = 1200

OUTPUT_FILE = "/content/01_sentinel2_rgb_amazonas_real.png"

# ------------------------------------------------------------
# 2. SERVICIO WMS DE SENTINEL-2 CLOUDLESS
# ------------------------------------------------------------

WMS_URL = (
    "https://tiles.maps.eox.at/wms?"
    "SERVICE=WMS&"
    "VERSION=1.1.1&"
    "REQUEST=GetMap&"
    "LAYERS=s2cloudless-2025&"
    "STYLES=&"
    "SRS=EPSG:4326&"
    f"BBOX={BBOX}&"
    f"WIDTH={WIDTH}&"
    f"HEIGHT={HEIGHT}&"
    "FORMAT=image/jpeg"
)

print("Descargando imagen Sentinel-2 real...")
print("Área: Chachapoyas y alrededores, Amazonas, Perú")

# ------------------------------------------------------------
# 3. DESCARGAR LA IMAGEN
# ------------------------------------------------------------

response = requests.get(WMS_URL, timeout=120)

if response.status_code != 200:
    raise RuntimeError(
        f"No se pudo descargar la imagen. "
        f"Código HTTP: {response.status_code}\n"
        f"Respuesta: {response.text[:300]}"
    )

content_type = response.headers.get("content-type", "")

if "image" not in content_type:
    raise RuntimeError(
        "El servidor no devolvió una imagen.\n"
        f"Tipo recibido: {content_type}\n"
        f"Respuesta: {response.text[:500]}"
    )

satellite_image = Image.open(BytesIO(response.content)).convert("RGB")

print("✅ Imagen descargada correctamente.")
print("Dimensiones originales:", satellite_image.size)

# ------------------------------------------------------------
# 4. CREAR ESPACIO PARA TÍTULO Y FUENTE
# ------------------------------------------------------------

header_height = 115
footer_height = 75

canvas = Image.new(
    "RGB",
    (
        satellite_image.width,
        satellite_image.height + header_height + footer_height
    ),
    "white"
)

canvas.paste(satellite_image, (0, header_height))

draw = ImageDraw.Draw(canvas)

# Utilizar fuentes disponibles en Colab
try:
    title_font = ImageFont.truetype(
        "/usr/share/fonts/truetype/dejavu/DejaVuSans-Bold.ttf",
        38
    )

    subtitle_font = ImageFont.truetype(
        "/usr/share/fonts/truetype/dejavu/DejaVuSans.ttf",
        25
    )

    source_font = ImageFont.truetype(
        "/usr/share/fonts/truetype/dejavu/DejaVuSans.ttf",
        20
    )

except Exception:
    title_font = ImageFont.load_default()
    subtitle_font = ImageFont.load_default()
    source_font = ImageFont.load_default()

# ------------------------------------------------------------
# 5. AGREGAR TÍTULO Y ATRIBUCIÓN
# ------------------------------------------------------------

title = "Sentinel-2 RGB — Amazonas, Perú"
subtitle = "Mosaico casi libre de nubes 2025 | Chachapoyas y alrededores"

source = (
    "Fuente: Sentinel-2 Cloudless 2025 — EOX IT Services GmbH. "
    "Contiene datos Copernicus Sentinel modificados."
)

draw.text(
    (35, 20),
    title,
    fill="black",
    font=title_font
)

draw.text(
    (35, 68),
    subtitle,
    fill=(70, 70, 70),
    font=subtitle_font
)

draw.text(
    (35, satellite_image.height + header_height + 20),
    source,
    fill=(60, 60, 60),
    font=source_font
)

# ------------------------------------------------------------
# 6. GUARDAR COMO PNG
# ------------------------------------------------------------

canvas.save(
    OUTPUT_FILE,
    format="PNG",
    optimize=True
)

file_size_mb = Path(OUTPUT_FILE).stat().st_size / (1024 * 1024)

print("\n✅ Archivo guardado:")
print(OUTPUT_FILE)
print(f"Peso: {file_size_mb:.2f} MB")

# ------------------------------------------------------------
# 7. MOSTRAR LA IMAGEN
# ------------------------------------------------------------

plt.figure(figsize=(15, 11))
plt.imshow(canvas)
plt.axis("off")
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 8. DESCARGAR AUTOMÁTICAMENTE
# ------------------------------------------------------------

from google.colab import files

files.download(OUTPUT_FILE)
```
